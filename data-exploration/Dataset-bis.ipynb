## Dynamic data load

```python
dossier_dyslexic = "../data/2emeDataset/Dyslexic"
dossier_control = "../data/2emeDataset/Control"
```

```python
dyslexic_subject = [Path(f).stem for f in os.listdir(dossier_dyslexic) if f != 'Recording Data.csv']
control_subject = [Path(f).stem for f in os.listdir(dossier_control)]
```

```python
def get_subject_data(isDys: bool = True, idSubject: int = 0):
    dossier_dyslexic = "../data/2emeDataset/Dyslexic"
    dossier_control = "../data/2emeDataset/Control"

    dyslexic_subject = [Path(f).stem for f in os.listdir(dossier_dyslexic) if f.endswith(".csv") and f != "Recording Data.csv"]
    control_subject = [Path(f).stem for f in os.listdir(dossier_control) if f.endswith(".csv")]

    if isDys:
        if idSubject < len(dyslexic_subject):
            id = dyslexic_subject[idSubject]
            path = f'{dossier_dyslexic}/{id}.csv'
        else:
            print(f'The dyslexic subject with id: {idSubject} does not exist in the dataset')
            return None
    else :
        if idSubject < len(control_subject):
            id = control_subject[idSubject]
            path = f'{dossier_control}/{id}.csv'
        else:
            print(f'The control subject with id: {idSubject} does not exist in the dataset')
            return None

    # print(path)
    df = pd.read_csv(path, index_col=0)

    return {
            'id': id,
            'type': 'Dyslexic' if isDys else 'Control',
            'data': df,
        }
```

```python
def plot_subject_data(subject):
    fig, ax = plt.subplots(2, 2, figsize=(15, 10))
    fig.suptitle(f"{subject['type']} {subject['id']}")

    ax[0,0].set_title("Trajectories")
    ax[0,0].plot(subject["data"]["LX"], subject["data"]["LY"], label="Left (LX, LY)")
    ax[0,0].plot(subject["data"]["RX"], subject["data"]["RY"], label="Right (RX, RY)")
    ax[0,0].set_xlabel("X position")
    ax[0,0].set_ylabel("Y position")

    ax[0,1].set_title("Eyes X position / time")
    ax[0,1].plot(subject["data"]["T"], subject["data"]["RX"], label="")
    ax[0,1].plot(subject["data"]["T"], subject["data"]["LX"], label="")
    ax[0,1].set_xlabel("Time")

    ax[1,0].set_title("Left (x,y) position / time")
    ax[1,0].plot(subject["data"]["T"], subject["data"]["LX"], label="")
    ax[1,0].plot(subject["data"]["T"], subject["data"]["LY"], label="")
    ax[1,0].set_xlabel("Time")

    ax[1,1].set_title("Right (x,y) position / time")
    ax[1,1].plot(subject["data"]["T"], subject["data"]["RX"], label="")
    ax[1,1].plot(subject["data"]["T"], subject["data"]["RY"], label="")
    ax[1,1].set_xlabel("Time")
```

```python
my_subject = get_subject_data(True, random.randint(0, 87))

plot_subject_data(my_subject)
```
